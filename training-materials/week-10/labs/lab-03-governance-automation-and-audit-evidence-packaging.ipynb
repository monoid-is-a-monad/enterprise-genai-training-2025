# Lab 03 – Governance Automation & Audit Evidence Packaging

**Scenario:** The compliance team is performing an audit readiness check. You will automate evidence collection, prepare governance artifacts, and draft the executive briefing.

## Objectives
- Automate retrieval of policy, evaluation, and release evidence
- Populate the compliance audit runbook with latest procedures
- Build Responsible AI scorecard snapshot for governance review
- Draft executive/stakeholder briefing using Week 10 outcomes

```python
from pathlib import Path

required = [
    Path('resources/compliance-audit-runbook.md'),
    Path('resources/responsible-ai-scorecard.csv'),
    Path('docs/governance/'),
    Path('artifacts/'),
]
missing = [str(p) for p in required if not p.exists()]
if missing:
    raise FileNotFoundError(f'Set up governance artifacts before running lab: {missing}')

print('Governance artifacts detected. Proceed with automation.')
```

## Step 1 – Evidence Collector Script
Aggregate policy, evaluation, and deployment metadata into a single manifest.

```python
import json
from datetime import datetime

manifest = {
    'generated_at': datetime.utcnow().isoformat(),
    'release_id': 'rel-2025w10-demo',
    'artifacts': {
        'policies': 'guardrails/policies/',
        'evaluations': 'artifacts/responsible-ai/',
        'deployments': 'artifacts/release-evidence/',
        'incidents': 'docs/incidents/',
    },
    'approvals': 'artifacts/policy-approval-log.md',
    'scorecard': 'resources/responsible-ai-scorecard.csv',
}
output_path = Path('artifacts/governance-evidence-manifest.json')
output_path.write_text(json.dumps(manifest, indent=2))
print(f'Evidence manifest generated at {output_path}')
```
