## 3. Dry Run Simulation
Simulate a dry run by invoking the pipeline with representative prompts and capturing timing.

```python
import time
import json
from orchestration.router import LLMRouter
from pipelines.poc import run_pipeline  # Ensure this module exports the PoC pipeline
\n
scenarios = [
    {"scene": "Hook", "prompt": "Summarize why our current risk triage is failing", "persona": "executive"},
    {"scene": "Deep Dive", "prompt": "Resolve case #4821 with guardrails", "persona": "analyst"},
    {"scene": "Impact", "prompt": "Show KPIs for last 24 hours", "persona": "executive"},
    {"scene": "CTA", "prompt": "Outline next steps for pilot go-live", "persona": "executive"},
    {"scene": "Contingency", "prompt": "Handle a prompt injection attempt", "persona": "analyst"}
],
router = LLMRouter(primary="gpt-4.1", fallback="gpt-4o-mini")
dry_run_log = []
\n
for scenario in scenarios:
    start = time.perf_counter()
    response = run_pipeline(scenario["prompt"], persona=scenario["persona"])
    duration_ms = (time.perf_counter() - start) * 1000
    dry_run_log.append({
        "scene": scenario["scene"],
        "persona": scenario["persona"],
        "latency_ms": round(duration_ms, 2),
        "excerpt": response.content[:200]
    })
\n
Path("reports").mkdir(exist_ok=True)
Path("reports/dry-run-log.json").write_text(json.dumps(dry_run_log, indent=2))
print("Dry run completed. Log stored at reports/dry-run-log.json")
```

Capture latency stats and ensure no scene exceeds 3 seconds.

## 5. Stakeholder Enablement Assets
Update or create the assets below so champions can retell the story:
- Executive one-pager with problem, solution, metrics, and next steps.
- FAQ with top objections and crisp responses.
- Cost summary and guardrail evidence snapshots.
- Demo toolkit folder link shared in stakeholder communication plan.

```python
from textwrap import dedent
\n
def draft_followup_email(next_steps: str, attachments: list[str]) -> str:
    """Generate a stakeholder follow-up email template."""
    attachments_line = "\n".join(f"- {item}" for item in attachments)
    return dedent(f"""\
Subject: Week 8 PoC Demo – Next Steps\n\n\
Hi team,\n\n\
Thank you for joining today's demo. Here are the key takeaways and proposed next steps:\n\n\
{next_steps}\n\n\
Artifacts:\n{attachments_line}\n\n\
Please reply by EOD tomorrow with any questions or approvals so we can lock the Week 9 plan.\n\n\
Best,\nDemo Lead\n""")
\n
email_body = draft_followup_email(
    next_steps="- Approve pilot scope\n- Provide data access for cohort B\n- Confirm budget envelope for Q1 rollout",
    attachments=[
        "Demo recording",
        "Executive one-pager",
        "Guardrail & readiness scorecard snapshot"
    ]
)
print(email_body)
```
